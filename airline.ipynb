```python
df["Days_left"] = (df["Days_left"] - days_left_min)/(days_left_max - days_left_min)
```

```python
# =========================================================
#         COMPLETE FLIGHT FARE PREDICTION SYSTEM
# =========================================================

import pandas as pd

# =========================================================
# MIN-MAX SCALING VALUES
# =========================================================

fare_min = 1307
fare_max = 143019

# =========================================================
# USER INPUTS
# =========================================================

# -----------------------------
# ORDINAL FEATURES
# -----------------------------

flight_class = int(input(
    "Class (Economy=0, Premium Economy=1, Business=2, First=3): "
))

total_stops = int(input(
    "Total Stops (non-stop=0, 1-stop=1, 2+-stop=2): "
))

# -----------------------------
# NUMERICAL FEATURES
# -----------------------------

duration_hours = float(input(
    "Duration in Hours (0.75 - 43.58): "
))

days_left = int(input(
    "Days Left Before Journey (1 - 50): "
))

journey_day = int(input(
    "Journey Date (1 - 31): "
))

journey_month = int(input(
    "Journey Month (1 - 12): "
))

is_weekend = int(input(
    "Is Weekend? (0=No, 1=Yes): "
))

# -----------------------------
# AIRLINE
# -----------------------------

print("\nAirlines:")
print("1 = Air India")
print("2 = AirAsia")
print("3 = AkasaAir")
print("4 = AllianceAir")
print("5 = GO FIRST")
print("6 = Indigo")
print("7 = SpiceJet")
print("8 = StarAir")
print("9 = Vistara")

airline = int(input("Choose Airline Number: "))

# -----------------------------
# SOURCE
# -----------------------------

print("\nSource Cities:")
print("1 = Ahmedabad")
print("2 = Bangalore")
print("3 = Chennai")
print("4 = Delhi")
print("5 = Hyderabad")
print("6 = Kolkata")
print("7 = Mumbai")

source = int(input("Choose Source City Number: "))

# -----------------------------
# DESTINATION
# -----------------------------

print("\nDestination Cities:")
print("1 = Ahmedabad")
print("2 = Bangalore")
print("3 = Chennai")
print("4 = Delhi")
print("5 = Hyderabad")
print("6 = Kolkata")
print("7 = Mumbai")

destination = int(input("Choose Destination City Number: "))

# -----------------------------
# DEPARTURE TIME
# -----------------------------

print("\nDeparture Time:")
print("1 = Before 6 AM")
print("2 = 6 AM - 12 PM")
print("3 = 12 PM - 6 PM")
print("4 = After 6 PM")

departure = int(input("Choose Departure Time Number: "))

# -----------------------------
# ARRIVAL TIME
# -----------------------------

print("\nArrival Time:")
print("1 = Before 6 AM")
print("2 = 6 AM - 12 PM")
print("3 = 12 PM - 6 PM")
print("4 = After 6 PM")

arrival = int(input("Choose Arrival Time Number: "))

# =========================================================
# CREATE INPUT DATAFRAME
# =========================================================

input_data = pd.DataFrame([{

    # Numerical / Ordinal
    "Class": flight_class,
    "Total_stops": total_stops,
    "Duration_in_hours": duration_hours,
    "Days_left": days_left,

    # Journey Days
    "Journey_day_Friday": 0,
    "Journey_day_Monday": 0,
    "Journey_day_Saturday": 0,
    "Journey_day_Sunday": 0,
    "Journey_day_Thursday": 0,
    "Journey_day_Tuesday": 0,
    "Journey_day_Wednesday": 0,

    # Airlines
    "Airline_Air India": 1 if airline == 1 else 0,
    "Airline_AirAsia": 1 if airline == 2 else 0,
    "Airline_AkasaAir": 1 if airline == 3 else 0,
    "Airline_AllianceAir": 1 if airline == 4 else 0,
    "Airline_GO FIRST": 1 if airline == 5 else 0,
    "Airline_Indigo": 1 if airline == 6 else 0,
    "Airline_SpiceJet": 1 if airline == 7 else 0,
    "Airline_StarAir": 1 if airline == 8 else 0,
    "Airline_Vistara": 1 if airline == 9 else 0,

    # Source
    "Source_Ahmedabad": 1 if source == 1 else 0,
    "Source_Bangalore": 1 if source == 2 else 0,
    "Source_Chennai": 1 if source == 3 else 0,
    "Source_Delhi": 1 if source == 4 else 0,
    "Source_Hyderabad": 1 if source == 5 else 0,
    "Source_Kolkata": 1 if source == 6 else 0,
    "Source_Mumbai": 1 if source == 7 else 0,

    # Destination
    "Destination_Ahmedabad": 1 if destination == 1 else 0,
    "Destination_Bangalore": 1 if destination == 2 else 0,
    "Destination_Chennai": 1 if destination == 3 else 0,
    "Destination_Delhi": 1 if destination == 4 else 0,
    "Destination_Hyderabad": 1 if destination == 5 else 0,
    "Destination_Kolkata": 1 if destination == 6 else 0,
    "Destination_Mumbai": 1 if destination == 7 else 0,

    # Departure
    "Departure_12 PM - 6 PM": 1 if departure == 3 else 0,
    "Departure_6 AM - 12 PM": 1 if departure == 2 else 0,
    "Departure_After 6 PM": 1 if departure == 4 else 0,
    "Departure_Before 6 AM": 1 if departure == 1 else 0,

    # Arrival
    "Arrival_12 PM - 6 PM": 1 if arrival == 3 else 0,
    "Arrival_6 AM - 12 PM": 1 if arrival == 2 else 0,
    "Arrival_After 6 PM": 1 if arrival == 4 else 0,
    "Arrival_Before 6 AM": 1 if arrival == 1 else 0,

    # Additional Features
    "journey_day": journey_day,
    "journey_month": journey_month,
    "is_weekend": is_weekend
}])

# =========================================================
# MATCH TRAINING COLUMN ORDER
# =========================================================

input_data = input_data[x.columns]

# =========================================================
# PREDICT SCALED VALUE
# =========================================================

scaled_prediction = xgb.predict(input_data)[0]

# =========================================================
# MANUAL INVERSE MIN-MAX SCALING
# =========================================================

actual_fare = (
    scaled_prediction * (fare_max - fare_min)
) + fare_min

# =========================================================
# OUTPUT
# =========================================================

print("\n==========================================")
print("Predicted Flight Fare = ₹", round(actual_fare, 2))
print("==========================================")
```
